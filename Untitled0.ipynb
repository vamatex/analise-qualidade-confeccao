```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Carregue o seu arquivo CSV para o ambiente do Google Colab
# e substitua 'SEU_ARQUIVO.csv' pelo nome do arquivo.
file_path = '/content/SUFALU_495_LUCIANOAL_0341 (1).CSV'

try:
    # Tenta carregar o arquivo com o delimitador correto
    df = pd.read_csv(file_path, delimiter=';')
except Exception as e:
    print(f"Erro ao carregar o CSV: {e}")
    # Tenta com outra codificação se houver um erro
    df = pd.read_csv(file_path, delimiter=';', encoding='latin1')


# --- Limpeza e Preparação dos Dados ---
# Renomeia as colunas para facilitar o acesso
df.rename(columns={
    'Data': 'Date',
    'Ciclo': 'Cycle',
    'OP': 'ProductionOrder',
    'Nivel': 'Level',
    'REF': 'Reference',
    'Class 4': 'Classification',
    'Qtd Real': 'TotalQuantity',
    '1ª': 'FirstQuality',
    '2º': 'SecondQuality',
    'Sac': 'Scrap',
    '%2ª': 'PercentSecondQuality',
    '%Sac': 'PercentScrap',
    'Fornecedor': 'Supplier'
}, inplace=True)

# Converte colunas numéricas, trocando vírgula por ponto
numeric_cols = ['TotalQuantity', 'FirstQuality', 'SecondQuality', 'Scrap', 'PercentSecondQuality', 'PercentScrap']
for col in numeric_cols:
    if df[col].dtype == 'object':
        df[col] = pd.to_numeric(df[col].str.replace(',', '.'), errors='coerce')

# Preenche valores nulos nas colunas de qualidade com 0
quality_cols = ['FirstQuality', 'SecondQuality', 'Scrap']
df[quality_cols] = df[quality_cols].fillna(0)


# --- Análises ---

# 1. Análise de Fornecedores
supplier_summary = df.groupby('Supplier')[['FirstQuality', 'SecondQuality', 'Scrap']].sum().reset_index()

# Top 10 fornecedores para cada categoria
top_10_first_quality = supplier_summary.nlargest(10, 'FirstQuality')
top_10_second_quality = supplier_summary.nlargest(10, 'SecondQuality')
top_10_scrap = supplier_summary.nlargest(10, 'Scrap')

# 2. Comparativo de Qualidade
supplier_summary['TotalProduction'] = supplier_summary['FirstQuality'] + supplier_summary['SecondQuality'] + supplier_summary['Scrap']
supplier_summary['PercentSecondQuality'] = (supplier_summary['SecondQuality'] / supplier_summary['TotalProduction']) * 100
supplier_summary['PercentScrap'] = (supplier_summary['Scrap'] / supplier_summary['TotalProduction']) * 100
supplier_quality_comparison = supplier_summary.sort_values(by=['PercentSecondQuality', 'PercentScrap'], ascending=False)

# 3. Análise de Produtos (Referências)
product_second_quality = df.groupby(['Supplier', 'Reference'])['SecondQuality'].sum().reset_index()
top_20_second_quality_products = product_second_quality.nlargest(20, 'SecondQuality')

# Encontra referências fornecidas por mais de um fornecedor
common_references = df.groupby('Reference')['Supplier'].nunique()
common_references = common_references[common_references > 1].index.tolist()

# Análise de qualidade para referências comuns
common_ref_df = df[df['Reference'].isin(common_references)]
common_ref_quality = common_ref_df.groupby(['Reference', 'Supplier'])[['FirstQuality', 'SecondQuality', 'Scrap']].sum().reset_index()
common_ref_quality['TotalProduction'] = common_ref_quality['FirstQuality'] + common_ref_quality['SecondQuality'] + common_ref_quality['Scrap']
common_ref_quality['PercentSecondQuality'] = (common_ref_quality['SecondQuality'] / common_ref_quality['TotalProduction']) * 100
common_ref_quality['PercentScrap'] = (common_ref_quality['Scrap'] / common_ref_quality['TotalProduction']) * 100


# --- Geração de Arquivos ---

# Arquivo Excel
with pd.ExcelWriter('analise_fornecedores.xlsx') as writer:
    supplier_summary.to_excel(writer, sheet_name='Resumo_Fornecedores', index=False)
    top_10_first_quality.to_excel(writer, sheet_name='Top_10_Primeira_Qualidade', index=False)
    top_10_second_quality.to_excel(writer, sheet_name='Top_10_Segunda_Qualidade', index=False)
    top_10_scrap.to_excel(writer, sheet_name='Top_10_Sacrificado', index=False)
    supplier_quality_comparison.to_excel(writer, sheet_name='Comparativo_Qualidade', index=False)
    top_20_second_quality_products.to_excel(writer, sheet_name='Top_20_Produtos_2a_Qualidade', index=False)
    common_ref_quality.to_excel(writer, sheet_name='Qualidade_Refs_Comuns', index=False)

print("Arquivo 'analise_fornecedores.xlsx' gerado com sucesso!")


# --- Visualizações ---

# Top 10 Fornecedores - Primeira Qualidade
plt.figure(figsize=(12, 8))
sns.barplot(data=top_10_first_quality.sort_values('FirstQuality', ascending=False), x='FirstQuality', y='Supplier', palette='viridis')
plt.title('Top 10 Fornecedores por Quantidade de Peças de Primeira Qualidade')
plt.xlabel('Quantidade de Primeira Qualidade')
plt.ylabel('Fornecedor')
plt.tight_layout()
plt.savefig('top_fornecedores_primeira_qualidade.png')
plt.show()

# Top 10 Fornecedores - Segunda Qualidade
plt.figure(figsize=(12, 8))
sns.barplot(data=top_10_second_quality.sort_values('SecondQuality', ascending=False), x='SecondQuality', y='Supplier', palette='plasma')
plt.title('Top 10 Fornecedores por Quantidade de Peças de Segunda Qualidade')
plt.xlabel('Quantidade de Segunda Qualidade')
plt.ylabel('Fornecedor')
plt.tight_layout()
plt.savefig('top_fornecedores_segunda_qualidade.png')
plt.show()

# Top 10 Fornecedores - Sacrifício
plt.figure(figsize=(12, 8))
sns.barplot(data=top_10_scrap.sort_values('Scrap', ascending=False), x='Scrap', y='Supplier', palette='magma')
plt.title('Top 10 Fornecedores por Quantidade de Peças de Sacrifício')
plt.xlabel('Quantidade de Sacrifício')
plt.ylabel('Fornecedor')
plt.tight_layout()
plt.savefig('top_fornecedores_sacrificio.png')
plt.show()

# Top 20 produtos com mais peças de segunda qualidade
plt.figure(figsize=(12, 10))
sns.barplot(data=top_20_second_quality_products, x='SecondQuality', y='Reference', hue='Supplier', dodge=False)
plt.title('Top 20 Referências com Mais Peças de Segunda Qualidade')
plt.xlabel('Quantidade de Segunda Qualidade')
plt.ylabel('Referência')
plt.legend(title='Fornecedor', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.savefig('top_produtos_segunda_qualidade.png')
plt.show()

print("Gráficos gerados e salvos como imagens PNG.")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Carregue o seu arquivo CSV para o ambiente do Google Colab
# e substitua pelo nome do arquivo, se necessário.
file_path = '/content/SUFALU_495_LUCIANOAL_0341 (1).CSV'

try:
    df = pd.read_csv(file_path, delimiter=';')
except Exception as e:
    # Tenta com outra codificação se houver um erro
    df = pd.read_csv(file_path, delimiter=';', encoding='latin1')

# --- Anonimização dos Dados ---

# Anonimiza os nomes dos Fornecedores
unique_suppliers = df['Fornecedor'].unique()
supplier_mapping = {supplier: f"Fornecedor {i+1}" for i, supplier in enumerate(unique_suppliers)}
df['Fornecedor'] = df['Fornecedor'].map(supplier_mapping)

# Anonimiza as Referências (Produtos)
unique_references = df['REF'].unique()
reference_mapping = {ref: f"Produto {i+1}" for i, ref in enumerate(unique_references)}
df['REF'] = df['REF'].map(reference_mapping)


# --- Limpeza e Preparação dos Dados (igual à análise anterior) ---
df.rename(columns={
    'Data': 'Date', 'Ciclo': 'Cycle', 'OP': 'ProductionOrder',
    'Nivel': 'Level', 'REF': 'Reference', 'Class 4': 'Classification',
    'Qtd Real': 'TotalQuantity', '1ª': 'FirstQuality', '2º': 'SecondQuality',
    'Sac': 'Scrap', '%2ª': 'PercentSecondQuality', '%Sac': 'PercentScrap',
    'Fornecedor': 'Supplier'
}, inplace=True)

numeric_cols = ['TotalQuantity', 'FirstQuality', 'SecondQuality', 'Scrap', 'PercentSecondQuality', 'PercentScrap']
for col in numeric_cols:
    if df[col].dtype == 'object':
        df[col] = pd.to_numeric(df[col].str.replace(',', '.'), errors='coerce')

df[['FirstQuality', 'SecondQuality', 'Scrap']] = df[['FirstQuality', 'SecondQuality', 'Scrap']].fillna(0)

# --- Reexecução da Análise com Dados Anonimizados ---

# 1. Análise de Fornecedores
supplier_summary = df.groupby('Supplier')[['FirstQuality', 'SecondQuality', 'Scrap']].sum().reset_index()
top_10_first_quality = supplier_summary.nlargest(10, 'FirstQuality')
top_10_second_quality = supplier_summary.nlargest(10, 'SecondQuality')
top_10_scrap = supplier_summary.nlargest(10, 'Scrap')

# 2. Comparativo de Qualidade
supplier_summary['TotalProduction'] = supplier_summary['FirstQuality'] + supplier_summary['SecondQuality'] + supplier_summary['Scrap']
supplier_summary['PercentSecondQuality'] = (supplier_summary['SecondQuality'] / supplier_summary['TotalProduction']) * 100
supplier_summary['PercentScrap'] = (supplier_summary['Scrap'] / supplier_summary['TotalProduction']) * 100
supplier_quality_comparison = supplier_summary.sort_values(by=['PercentSecondQuality', 'PercentScrap'], ascending=False)

# 3. Análise de Produtos
product_second_quality = df.groupby(['Supplier', 'Reference'])['SecondQuality'].sum().reset_index()
top_20_second_quality_products = product_second_quality.nlargest(20, 'SecondQuality')

# --- Geração dos Arquivos Anonimizados ---

# Novo Arquivo Excel Anonimizado
excel_filename = 'analise_anonimizada_fornecedores.xlsx'
with pd.ExcelWriter(excel_filename) as writer:
    supplier_summary.to_excel(writer, sheet_name='Resumo_Fornecedores', index=False)
    top_10_first_quality.to_excel(writer, sheet_name='Top_10_Primeira_Qualidade', index=False)
    top_10_second_quality.to_excel(writer, sheet_name='Top_10_Segunda_Qualidade', index=False)
    top_10_scrap.to_excel(writer, sheet_name='Top_10_Sacrificado', index=False)
    supplier_quality_comparison.to_excel(writer, sheet_name='Comparativo_Qualidade', index=False)
    top_20_second_quality_products.to_excel(writer, sheet_name='Top_20_Produtos_2a_Qualidade', index=False)

print(f"Arquivo '{excel_filename}' gerado com sucesso!")


# --- Novas Visualizações Anonimizadas ---

plt.figure(figsize=(12, 8))
sns.barplot(data=top_10_first_quality.sort_values('FirstQuality', ascending=False), x='FirstQuality', y='Supplier', palette='viridis')
plt.title('Top 10 Fornecedores (Anônimos) por Peças de Primeira Qualidade')
plt.xlabel('Quantidade de Primeira Qualidade')
plt.ylabel('Fornecedor Fictício')
plt.tight_layout()
plt.savefig('anon_top_fornecedores_primeira.png')
plt.show()

plt.figure(figsize=(12, 8))
sns.barplot(data=top_10_second_quality.sort_values('SecondQuality', ascending=False), x='SecondQuality', y='Supplier', palette='plasma')
plt.title('Top 10 Fornecedores (Anônimos) por Peças de Segunda Qualidade')
plt.xlabel('Quantidade de Segunda Qualidade')
plt.ylabel('Fornecedor Fictício')
plt.tight_layout()
plt.savefig('anon_top_fornecedores_segunda.png')
plt.show()

plt.figure(figsize=(12, 8))
sns.barplot(data=top_10_scrap.sort_values('Scrap', ascending=False), x='Scrap', y='Supplier', palette='magma')
plt.title('Top 10 Fornecedores (Anônimos) por Peças de Sacrifício')
plt.xlabel('Quantidade de Sacrifício')
plt.ylabel('Fornecedor Fictício')
plt.tight_layout()
plt.savefig('anon_top_fornecedores_sacrificio.png')
plt.show()

plt.figure(figsize=(12, 10))
sns.barplot(data=top_20_second_quality_products, x='SecondQuality', y='Reference', hue='Supplier', dodge=False)
plt.title('Top 20 Produtos (Anônimos) com Mais Peças de Segunda Qualidade')
plt.xlabel('Quantidade de Segunda Qualidade')
plt.ylabel('Produto Fictício')
plt.legend(title='Fornecedor', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.savefig('anon_top_produtos_segunda.png')
plt.show()

print("Gráficos anonimizados gerados e salvos como imagens PNG.")
```

```python
import pandas as pd

# O nome do arquivo DEVE ser exatamente este no ambiente do Colab
file_path = '/content/SUFALU_495_LUCIANOAL_0341 (1).CSV'

try:
    # Tenta carregar o arquivo com a codificação correta
    df = pd.read_csv(file_path, delimiter=';', encoding='latin1')

    print("Arquivo carregado com sucesso! Iniciando a análise e anonimização...")

    # --- Anonimização ---
    unique_suppliers = df['Fornecedor'].unique()
    supplier_mapping = {supplier: f"Fornecedor {i+1}" for i, supplier in enumerate(unique_suppliers)}
    df['Fornecedor'] = df['Fornecedor'].map(supplier_mapping)

    unique_references = df['REF'].unique()
    reference_mapping = {ref: f"Produto {i+1}" for i, ref in enumerate(unique_references)}
    df['REF'] = df['REF'].map(reference_mapping)


    # --- Limpeza e Análise ---
    df.rename(columns={
        'Data': 'Date', 'REF': 'Reference', 'Qtd Real': 'TotalQuantity',
        '1ª': 'FirstQuality', '2º': 'SecondQuality', 'Sac': 'Scrap',
        'Fornecedor': 'Supplier'
    }, inplace=True)

    numeric_cols = ['TotalQuantity', 'FirstQuality', 'SecondQuality', 'Scrap']
    for col in numeric_cols:
        if df[col].dtype == 'object':
            df[col] = pd.to_numeric(df[col].str.replace(',', '.'), errors='coerce')

    df[numeric_cols] = df[numeric_cols].fillna(0)

    # 1. Análise Geral de Fornecedores
    supplier_summary = df.groupby('Supplier')[['FirstQuality', 'SecondQuality', 'Scrap']].sum().reset_index()
    top_10_first_quality = supplier_summary.nlargest(10, 'FirstQuality')
    top_10_second_quality = supplier_summary.nlargest(10, 'SecondQuality')
    top_10_scrap = supplier_summary.nlargest(10, 'Scrap')

    # 2. Comparativo de Qualidade Percentual
    supplier_summary['TotalProduction'] = supplier_summary['FirstQuality'] + supplier_summary['SecondQuality'] + supplier_summary['Scrap']
    supplier_summary['PercentSecondQuality'] = 0
    supplier_summary.loc[supplier_summary['TotalProduction'] > 0, 'PercentSecondQuality'] = (supplier_summary['SecondQuality'] / supplier_summary['TotalProduction']) * 100
    supplier_summary['PercentScrap'] = 0
    supplier_summary.loc[supplier_summary['TotalProduction'] > 0, 'PercentScrap'] = (supplier_summary['Scrap'] / supplier_summary['TotalProduction']) * 100
    supplier_quality_comparison = supplier_summary.sort_values(by=['PercentSecondQuality', 'PercentScrap'], ascending=False)

    # 3. Análise de Produtos (Referências)
    product_second_quality = df.groupby(['Supplier', 'Reference'])['SecondQuality'].sum().reset_index()
    top_20_second_quality_products = product_second_quality.nlargest(20, 'SecondQuality')

    # Análise de Referências Comuns
    common_references = df.groupby('Reference')['Supplier'].nunique()
    common_references = common_references[common_references > 1].index.tolist()
    common_ref_df = df[df['Reference'].isin(common_references)]
    common_ref_quality = common_ref_df.groupby(['Reference', 'Supplier'])[['FirstQuality', 'SecondQuality', 'Scrap']].sum().reset_index()

    if not common_ref_quality.empty:
      common_ref_quality['TotalProduction'] = common_ref_quality['FirstQuality'] + common_ref_quality['SecondQuality'] + common_ref_quality['Scrap']
      common_ref_quality['PercentSecondQuality'] = 0
      common_ref_quality.loc[common_ref_quality['TotalProduction'] > 0, 'PercentSecondQuality'] = (common_ref_quality['SecondQuality'] / common_ref_quality['TotalProduction']) * 100
      common_ref_quality['PercentScrap'] = 0
      common_ref_quality.loc[common_ref_quality['TotalProduction'] > 0, 'PercentScrap'] = (common_ref_quality['Scrap'] / common_ref_quality['TotalProduction']) * 100

    # --- Geração do Arquivo Excel Anonimizado ---
    excel_filename = 'analise_anonimizada_fornecedores.xlsx'
    with pd.ExcelWriter(excel_filename) as writer:
        supplier_summary.to_excel(writer, sheet_name='Resumo_Fornecedores', index=False)
        top_10_first_quality.to_excel(writer, sheet_name='Top_10_Primeira_Qualidade', index=False)
        top_10_second_quality.to_excel(writer, sheet_name='Top_10_Segunda_Qualidade', index=False)
        top_10_scrap.to_excel(writer, sheet_name='Top_10_Sacrificado', index=False)
        supplier_quality_comparison.to_excel(writer, sheet_name='Comparativo_Qualidade', index=False)
        top_20_second_quality_products.to_excel(writer, sheet_name='Top_20_Produtos_2a_Qualidade', index=False)
        if not common_ref_quality.empty:
            common_ref_quality.to_excel(writer, sheet_name='Qualidade_Refs_Comuns', index=False)

    print(f"\nArquivo '{excel_filename}' gerado com sucesso!")
    print("Você já pode fazer o download dele no menu de arquivos à esquerda.")

except FileNotFoundError:
    print(f"ERRO: O arquivo '{file_path}' não foi encontrado.")
    print("Por favor, verifique se você fez o upload do arquivo e se o nome corresponde exatamente.")
except Exception as e:
    print(f"Ocorreu um erro inesperado: {e}")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Carregue seu arquivo CSV para o ambiente do Colab
file_path = '/content/SUFALU_495_LUCIANOAL_0341 (1).CSV'

try:
    df = pd.read_csv(file_path, delimiter=';', encoding='latin1')

    # --- Anonimização ---
    unique_suppliers = df['Fornecedor'].unique()
    supplier_mapping = {supplier: f"Fornecedor {i+1}" for i, supplier in enumerate(unique_suppliers)}
    df['Fornecedor'] = df['Fornecedor'].map(supplier_mapping)

    unique_references = df['REF'].unique()
    reference_mapping = {ref: f"Produto {i+1}" for i, ref in enumerate(unique_references)}
    df['REF'] = df['REF'].map(reference_mapping)

    # Salvar o arquivo de dados anonimizados para o GitHub
    df.to_csv('dados_anonimizados.csv', index=False, sep=';')
    print("Arquivo 'dados_anonimizados.csv' criado com sucesso!")

    # --- Limpeza e Análise ---
    df.rename(columns={
        'REF': 'Reference', 'Qtd Real': 'TotalQuantity', '1ª': 'FirstQuality',
        '2º': 'SecondQuality', 'Sac': 'Scrap', 'Fornecedor': 'Supplier'
    }, inplace=True)

    numeric_cols = ['TotalQuantity', 'FirstQuality', 'SecondQuality', 'Scrap']
    for col in numeric_cols:
        if df[col].dtype == 'object':
            df[col] = pd.to_numeric(df[col].str.replace(',', '.'), errors='coerce')
    df[numeric_cols] = df[numeric_cols].fillna(0)

    supplier_summary = df.groupby('Supplier')[['FirstQuality', 'SecondQuality', 'Scrap']].sum()
    top_10_first_quality = supplier_summary.nlargest(10, 'FirstQuality').reset_index()
    top_10_second_quality = supplier_summary.nlargest(10, 'SecondQuality').reset_index()
    top_10_scrap = supplier_summary.nlargest(10, 'Scrap').reset_index()

    product_second_quality = df.groupby(['Supplier', 'Reference'])['SecondQuality'].sum().reset_index()
    top_20_second_quality_products = product_second_quality.nlargest(20, 'SecondQuality')

    # --- Geração dos Gráficos ---
    # Gráfico 1
    plt.figure(figsize=(12, 8))
    sns.barplot(data=top_10_first_quality.sort_values('FirstQuality', ascending=False), x='FirstQuality', y='Supplier', palette='viridis')
    plt.title('Top 10 Fornecedores por Peças de Primeira Qualidade')
    plt.xlabel('Quantidade de Primeira Qualidade')
    plt.ylabel('Fornecedor (Anônimo)')
    plt.tight_layout()
    plt.savefig('grafico_01_primeira_qualidade.png')
    plt.close()

    # Gráfico 2
    plt.figure(figsize=(12, 8))
    sns.barplot(data=top_10_second_quality.sort_values('SecondQuality', ascending=False), x='SecondQuality', y='Supplier', palette='plasma')
    plt.title('Top 10 Fornecedores por Peças de Segunda Qualidade')
    plt.xlabel('Quantidade de Segunda Qualidade')
    plt.ylabel('Fornecedor (Anônimo)')
    plt.tight_layout()
    plt.savefig('grafico_02_segunda_qualidade.png')
    plt.close()

    # Gráfico 3
    plt.figure(figsize=(12, 8))
    sns.barplot(data=top_10_scrap.sort_values('Scrap', ascending=False), x='Scrap', y='Supplier', palette='magma')
    plt.title('Top 10 Fornecedores por Peças de Sacrifício (Refugo)')
    plt.xlabel('Quantidade de Sacrifício')
    plt.ylabel('Fornecedor (Anônimo)')
    plt.tight_layout()
    plt.savefig('grafico_03_sacrificio.png')
    plt.close()

    # Gráfico 4
    plt.figure(figsize=(12, 10))
    sns.barplot(data=top_20_second_quality_products, x='SecondQuality', y='Reference', hue='Supplier', dodge=False)
    plt.title('Top 20 Produtos com Mais Peças de Segunda Qualidade')
    plt.xlabel('Quantidade de Segunda Qualidade')
    plt.ylabel('Produto (Anônimo)')
    plt.legend(title='Fornecedor')
    plt.tight_layout()
    plt.savefig('grafico_04_produtos_segunda_qualidade.png')
    plt.close()

    print("Gráficos gerados com sucesso!")

except FileNotFoundError:
    print(f"ERRO: Arquivo '{file_path}' não encontrado. Faça o upload antes de executar.")
except Exception as e:
    print(f"Ocorreu um erro: {e}")
```
